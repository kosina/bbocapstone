```python
##############################################################
# Import the libraries that I will need for this exercise
##############################################################

from sklearn.datasets import load_wine
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report
```

```python
##############################################################
# Load the Wine dataset
# Here I separate the input variables (X) from the labels (y)
##############################################################

data = load_wine()

X = data.data
y = data.target
```

```python
##############################################################
# Split the data
#
# The aim is to obtain:
# - 60% training data
# - 20% validation data
# - 20% test data
#
# As train_test_split() only creates two groups,
# I need to do the split in two different steps.
##############################################################

# First I keep 20% of the data as the final test set.
# The remaining 80% will be used for training and validation.
X_train_val, X_test, y_train_val, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

# Now I split the remaining 80%.
# I take 25% of it for validation.
# This corresponds to 20% of the original dataset.
X_train, X_val, y_train, y_val = train_test_split(
    X_train_val,
    y_train_val,
    test_size=0.25,
    random_state=42,
    stratify=y_train_val
)
```

```python
##############################################################
# Create and train the Logistic Regression model
##############################################################

model = LogisticRegression(
    max_iter=5000,
    random_state=42
)

# Train the model using only the training data
model.fit(X_train, y_train)
```

```python
##############################################################
# Evaluate the model using the validation set
#
# I first make predictions and then calculate
# the validation accuracy.
##############################################################

val_preds = model.predict(X_val)

val_accuracy = accuracy_score(y_val, val_preds)

print(f"Validation Accuracy: {val_accuracy:.4f}")
```

```python
##############################################################
# Evaluate the model on the test set
#
# This gives an estimate of how well the model
# performs on unseen data.
##############################################################

test_preds = model.predict(X_test)

test_accuracy = accuracy_score(y_test, test_preds)

print(f"Test Accuracy: {test_accuracy:.4f}")
```

```python
##############################################################
# Print a more detailed summary of the results
#
# This report includes precision, recall,
# F1-score and support for each class.
##############################################################

print("\nClassification Report on Test Set:")
print(
    classification_report(
        y_test,
        test_preds,
        target_names=data.target_names
    )
)
```

```python
##############################################################
# K-Fold Cross-Validation on the Wine dataset
##############################################################

from sklearn.model_selection import StratifiedKFold, cross_val_score, cross_val_predict
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report
import numpy as np

# Step 1: Hold out a final test set (this stays untouched until the very end)
X_train_val, X_test, y_train_val, y_test = train_test_split(
    X, y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

##############################################################
# Step 2: Set up K-Fold cross-validation on the remaining data
#
# StratifiedKFold keeps class proportions balanced in every fold,
# which matters for a dataset this small.
##############################################################

k = 5
skf = StratifiedKFold(n_splits=k, shuffle=True, random_state=42)

model = LogisticRegression(max_iter=5000, random_state=42)

# cross_val_score automatically: splits data into k folds,
# trains on k-1 folds, validates on the remaining fold,
# and repeats k times
cv_scores = cross_val_score(model, X_train_val, y_train_val, cv=skf, scoring='accuracy')

print(f"Cross-validation accuracy per fold: {np.round(cv_scores, 4)}")
print(f"Mean CV accuracy: {cv_scores.mean():.4f}")
print(f"Std Dev of CV accuracy: {cv_scores.std():.4f}")

##############################################################
# Step 3: Train the final model on ALL train_val data
# (now that CV has given us confidence in the model/settings)
##############################################################

final_model = LogisticRegression(max_iter=5000, random_state=42)
final_model.fit(X_train_val, y_train_val)

##############################################################
# Step 4: Evaluate on the untouched test set
##############################################################

test_preds = final_model.predict(X_test)
test_accuracy = accuracy_score(y_test, test_preds)

print(f"\nTest Accuracy: {test_accuracy:.4f}")

print("\nClassification Report on Test Set:")
print(classification_report(y_test, test_preds, target_names=data.target_names))
```
